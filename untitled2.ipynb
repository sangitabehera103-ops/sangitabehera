```python
import streamlit as st
import numpy as np
from PIL import Image
import importlib

st.set_page_config(page_title="Image Prediction App", page_icon="🖼️", layout="centered")

# Defer importing tensorflow until runtime to avoid static import errors in
# environments where tensorflow isn't installed. load_model will attempt to
# import it when needed.
tf = None

st.title("🖼️ Image Prediction Web App")
st.write("Upload an image and click Predict.")

@st.cache_resource
def load_model():
    global tf
    if tf is None:
        try:
            tf = importlib.import_module("tensorflow")
        except Exception:
            return None
    try:
        return tf.keras.models.load_model("model.keras")
    except Exception:
        return None

model = load_model()

uploaded_file = st.file_uploader("Choose an image", type=["jpg", "jpeg", "png"])

if uploaded_file is not None:
    image = Image.open(uploaded_file).convert("RGB")
    st.image(image, caption="Uploaded Image", use_container_width=True)

    img = image.resize((224, 224))
    img = np.array(img) / 255.0
    img = np.expand_dims(img, axis=0)

    if st.button("Predict"):
        if model is None:
            st.error("Model unavailable. Install TensorFlow and ensure model.keras is present.")
        else:
            predictions = model.predict(img)
            predicted_class = np.argmax(predictions, axis=1)[0]
            confidence = np.max(predictions)
            st.write(f"Predicted Class: {predicted_class}")
            st.info(f"Confidence: {confidence*100:.2f}%")
```

Finally, run the Streamlit app. `pyngrok` will create a public URL that you can click to interact with your app. Make sure to get an `ngrok` authentication token from [ngrok.com](https://dashboard.ngrok.com/get-started/your-authtoken) and add it to Colab's secrets manager (🔑 icon in the left panel) with the name `NGROK_AUTH_TOKEN`.

```python
from pyngrok import ngrok
from google.colab import userdata
import os

# Set your ngrok authtoken. Get it from https://dashboard.ngrok.com/get-started/your-authtoken
NGROK_AUTH_TOKEN = userdata.get('NGROK_AUTH_TOKEN')
ngrok.set_auth_token(NGROK_AUTH_TOKEN)

# Run Streamlit in the background
os.system('nohup streamlit run app.py &')

# Open a ngrok tunnel to the Streamlit port (8501 by default)
pub_url = ngrok.connect(addr='8501')
print(f"Your Streamlit app is live at: {pub_url}")
```
